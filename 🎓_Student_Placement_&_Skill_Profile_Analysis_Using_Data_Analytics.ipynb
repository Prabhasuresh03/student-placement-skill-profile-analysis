# 🎓 Student Placement & Skill Profile Analysis

### Using Data Analytics
## 📌 Key Insights

1. The dataset contains 10,000 student records.

2. 1,659 students are placed, while 8,341 students are not placed.

3. The average CGPA of placed students is higher than that of not placed students.

4. Placed students have higher average communication skills than not placed students.

5. Placed students have completed more projects on average than not placed students.

6. Internship experience and academic performance were also analyzed to understand their relationship with placement.

## 📌 Conclusion

This analysis helps to understand the student factors associated with placement outcomes.
CGPA, communication skills, and projects completed show noticeable differences between placed and not placed students.

```python
import pandas as pd
df = pd.read_csv("college_student_placement_dataset.csv")
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
print("\nFirst 5 Rows:")
display(df.head())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

Check Duplicate Rows


```python
print("Duplicate Rows:", df.duplicated().sum())
```

Basic Statistics

```python
df.describe()
```

Placement Distribution

```python
print(df["Placement"].value_counts())
```

Placement Percentage

```python
print(df["Placement"].value_counts(normalize=True) * 100)
```

CGPA vs Placement

```python
df.groupby("Placement")["CGPA"].mean()
```

Internship

```python
df.groupby("Internship_Experience")["Placement"].value_counts()
```

Communication Skills

```python
df.groupby("Placement")["Communication_Skills"].mean()
```

Projects Completed

```python
df.groupby("Placement")["Projects_Completed"].mean()
```

Academic Performance

```python
df.groupby("Placement")["Academic_Performance"].mean()
```

CGPA vs Placement bar chart

```python
df.groupby("Placement")["CGPA"].mean().plot(kind="bar", title="Average CGPA by Placement")
```

Placement Distribution chart

```python
df["Placement"].value_counts().plot(
    kind="bar",
    title="Student Placement Distribution"
)
```

Chart 2 — Average CGPA by Placement

```python
df.groupby("Placement")["CGPA"].mean().plot(
    kind="bar",
    title="Average CGPA by Placement"
)
```

Chart 3 — Average Communication Skills by Placement

```python
df.groupby("Placement")["Communication_Skills"].mean().plot(
    kind="bar",
    title="Average Communication Skills by Placement"
)
```

Chart 4 — Average Projects Completed by Placement
